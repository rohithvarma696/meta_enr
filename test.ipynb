```python
import requests

API_KEY = 'eb0c0475'

def search_movies(movie_name):
    url = "http://www.omdbapi.com/"
    params = {
        "apikey": API_KEY,
        "s": movie_name  # search by title
    }
    response = requests.get(url, params=params)
    data = response.json()
    
    if data.get("Response") == "True":
        return data.get("Search", [])
    return []


def get_movie_details(imdb_id):
    url = "http://www.omdbapi.com/"
    params = {
        "apikey": API_KEY,
        "i": imdb_id  # fetch full details
    }
    response = requests.get(url, params=params)
    return response.json()


def find_movie_by_director(movie_name, director_name):
    search_results = search_movies(movie_name)
    
    matches = []
    
    for movie in search_results:
        imdb_id = movie.get("imdbID")
        details = get_movie_details(imdb_id)
        
        director = details.get("Director", "")
        
        # Match director (case-insensitive)
        if director_name.lower() in director.lower():
            matches.append(details)
    
    return matches


# Example usage
movie_name = "Patna wale dulhaniya le jayenge-Hindi"
director_name = "sanja"

results = find_movie_by_director(movie_name, director_name)

if results:
    print(results)
else:
    print("No matching movie found.")
```
